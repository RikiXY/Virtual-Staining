## Parsing e struttura dei parametri
Per quanto riguarda l'esecuzione e più precisamente la chiamata del programma da riga comando, lo script sfrutta il modulo `argparse` per gestire i parametri in ingresso.

Sono stati definiti tre argomenti principali:

- `path`: che specifica il percorso della cartella contenente le immagini _label-free_ e _stained_ (si consiglia di inserire tale cartella in Virtual-Staining/your_images_folder). **È l'unico argomento obbligatorio**
- `--seed SEED`: che specifica l'assenza o la presenza di un seed prestabilito. In caso affermativo si digiterà il numero del seed di seguito al flag (es: `--seed 42`). Questo approccio ci permette di garantire la riproducibilità dei risultati
- `--save_mask`: una flag booleana opzionale che indica se salvare o meno le maschere binarie delle sottoimmagini calcolate. La presenza di questa flag indica il salvataggio delle maschere.

A questi si aggiunge il l'argomento `--lang` che come detto prima consente di scegliere la lingua desiderata.

L'utilizzo di `argparse` permette di generare automaticamente una guida all'uso del programma, visualizzabile con il flag `-h` o `--help`.

Dal punto di vista pratico, la sintassi per eseguire lo script è la seguente: \
`python ollie_wan_kenobi.py path [--seed SEED] [--save_masks] [--lang {en, ita}]`
